```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("mountain.jpg")
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(img_gray, cmap="gray")
plt.axis("off")
```

```python
img_bill=cv2.bilateralFilter(img_gray,15,75,75)
plt.imshow(img_bill,cmap="gray")
plt.axis("off")
```

```python
img_blur=cv2.blur(img_gray, (5, 5))
plt.imshow(img_blur,cmap="gray")
plt.axis("off")
```

```python
img_medianBlur=cv2.medianBlur(img_gray, 5)
plt.imshow(img_medianBlur,cmap="gray")
plt.axis("off")
```

```python
img_GaussianBlur=cv2.GaussianBlur(img_gray, (5, 5), 0)
plt.imshow(img_medianBlur,cmap="gray")
plt.axis("off")
```
